```python

def is_integer(text):
    try:
        dato = int(text)
        if dato > 0:
            return True
        else:
            print("Entrada no válida. Introduzca un entero no negativo.")
            return False
    except:
        print("Entrada no válida. Introduzca un entero no negativo.")
        return False

def seconds_to_dhms(total_seconds):
    # 1 día = 86.400 segundos -> 24 horas x 60 minutos x 60 segundos 
    # 1 hora = 3600 segundos  
    # 1 min = 60 segundos 

    dias = total_seconds // 86400
    resto = total_seconds % 86400

    horas = resto // 3600
    resto = resto % 3600

    minutos = resto // 60
    segundos = resto % 60

    return (dias, horas, minutos, segundos)



def main():
    text = input("Introduzca los segundos: ")

    if is_integer(text):
        print("El valor introduciod es correcto.")
        d, h, m, s = seconds_to_dhms(int(text))
        print(f"{text} segundos son {d} d, {h} h, {m} min y {s} s.")

    else:
        main()

if __name__ == "__main__":
    main()







```

```python

def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

def leap_years_between(start, end):
    lista_bisiestos = []
    for year in range (start, end + 1, 1):
        if is_leap_year(year):
            lista_bisiestos.append(year)
    return lista_bisiestos


def explain_leap_year(year):
    if is_leap_year(year):
        if year % 4 == 0 and year % 100 != 0:
            return f"{year} es bisiesto: es divisible entre 4 y no entre 100."
        else:
            return f"{year} es bisiesto: es divisible entre 400."
    else:
        if year % 100 == 0:
            return f"{year} no es bisiesto: es divisible entre 100 pero no entre 400."
        else:
            return f"{year} no es bisiesto: no es divisible entre 4."

def main():
    año = int(input("Introduce un año: "))
    if año < 1582:
        print("El calendario gregoriano se estableció en 1582.")
    else:
        resultado = explain_leap_year(año)
        print(resultado)

    print("=*" * 60)
    
    # leap_years_between 
    inicio = 1896
    fin = 1912
    lista_bisiestos = leap_years_between(inicio, fin)
    print(f">>> leap_years_between({inicio}, {fin})\n{lista_bisiestos}")

if __name__ == "__main__":
    main()

```

```python
def shift_char(char, shift):
    if char.islower():
        nueva_pos = (ord(char) - ord('a') + shift) % 26
        return chr(ord('a') + nueva_pos)
    
    elif char.isupper():
        nueva_pos = (ord(char) - ord('A') + shift) % 26
        return chr(ord('A') + nueva_pos)
        
    else:
        return char

def encrypt(text, shift):
    lista = []
    for c in text:
        lista.append(shift_char(c, shift))
    return "".join(lista)

def decrypt(text, shift):
    lista = []
    for c in text:
        lista.append(shift_char(c, -shift))
    return "".join(lista)

def brute_force(cipher_text):
    lista = []
    print("\n","=*" * 40,"\nPosibles soluciones:\n")
    for i in range(26):
        lista.append(decrypt(cipher_text, i))
    return lista

def main():
    cadena1 = "Hola, Mundo!"
    cadena2 = "xyz"

    cadena1 = encrypt(cadena1, 3)
    print(cadena1)

    cadena1_1 = decrypt(cadena1, 3)
    print(cadena1_1)

    cadena2 = encrypt(cadena2, 3)
    print(cadena2)

    cadena2 = encrypt(cadena2, -3)
    print(cadena2) 

    cadena1 = brute_force(cadena1)
    print(cadena1)

if __name__ == "__main__":
    main()
```

```python
def normalize(text):
    lista = []
    for c in text:
        new_c = c.lower().replace('á', 'a').replace('é', 'e').replace('í', 'i').replace('ó', 'o').replace('ú', 'u').replace('ü', 'u')
        lista.append(new_c)
    return "".join(lista)
    
def is_palindrome(text):
    texto_limpio = normalize(text)
    texto_final = ""
    for c in texto_limpio:
        if c.isalnum():  
            texto_final += c

    return texto_final == texto_final[::-1]

def find_palindrome(sentences):
    lista = []
    for i in range(len(sentences)):
        if is_palindrome(sentences[i]):
            lista.append(sentences[i])
    return lista

def palindrome_words(text):
    lista = text.split()
    lista_palabras = find_palindrome(lista)
    print(lista_palabras)

def main():
    sentences = [
        "Anita lava la tina",
        "Dábale arroz a la zorra el abad",
        "Esto no es un palíndromo",
        "¿Acaso hubo búhos acá?",
    ]

    lista_frases = find_palindrome(sentences)
    print(lista_frases)
    palindrome_words("Ana vio un oso en el ojo de Anita")

if __name__ == "__main__":
    main()
```

```python
def is_prime(n):
    contador = 0
    for i in range (2, n):
        if n % i == 0:
            contador += 1
    return contador == 0 and n > 1

def prime_list(limit):
    lista_primos = []
    for i in range(2, limit + 1):
        if is_prime(i):
            lista_primos.append(i)
    return lista_primos

def check_palindrome(primes):
    lista_palin = []
    for i in primes:
        i = str(i)
        if i == i[::-1]:
            lista_palin.append(i)
    return lista_palin

def categorize_prime(prime):
    if prime < 10:
        return "p"
    if prime >= 10 and prime < 100:
        return "m"
    if prime >= 100:
        return "g"
    
def main():
    lista_numeros = [53, 1, 6, 34, 34, 11, 55, 17, 9, 67, 31, 22]
    lista_resultados = []

    for i in range(len(lista_numeros)):
        lista_resultados.append(is_prime(lista_numeros[i]))
    print(lista_resultados)

    lista_resultados = prime_list(150)
    print(lista_resultados)

    print(check_palindrome(lista_resultados))

    peq = []
    med = []
    grand = []
    for i in range(len(lista_resultados)):
        tam = categorize_prime(lista_resultados[i])
        if tam == "p":
            peq.append(lista_resultados[i])
        elif tam == "m":
            med.append(lista_resultados[i])
        if tam == "g":
            grand.append(lista_resultados[i])

    print(f"\nNúmeros pequeños: {peq}\nNúmeros medianos: {med}\nNúmeros grandes: {grand}")
            
if __name__ == "__main__":
    main()
















































```

```python
def remove_duplicates(numbers):
    lista_final = []
    for i in numbers:
        if i not in lista_final:
            lista_final.append(i)
    return lista_final

def rotate_left(numbers, k):
    k = k % len(numbers)
    return numbers[k:] + numbers[:k]

def split_even_odd(numbers): # Lista pares e impares
    pares, impares = [], []
    for i in numbers:
        if i % 2 == 0:
            pares.append(i)
        else: 
            impares.append(i)
    return [pares, impares]

def merge_sorted(a, b): # Recibe dos listas ya ordenadas de menor a mayor y
                        # devuelve una única lista ordenada. Recorra ambas listas a la vez con dos índices,
                        # sin ordenar el resultado final
    lista_ord = []
    i = 0
    j = 0

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            lista_ord.append(a[i])
            i += 1
        else:
            lista_ord.append(b[j])
            j += 1

    lista_ord.extend(a[i:]) # Los elementos que faltan por añadir
    lista_ord.extend(b[j:])

    return lista_ord

def second_largest(numbers): 
    
    if not numbers or len(numbers) < 2 or len(remove_duplicates(numbers)) == 1:
        return None

    numbers = remove_duplicates(numbers)
    numbers.sort() # Ordena de MENOR a MAYOR

    return numbers[-2]
    
def main():
    lista = [1, 2, 3, 4, 5, 6]

    print(remove_duplicates([3, 1, 3, 2, 1,]))
    print('*=' * 30)
    
    print(rotate_left(lista, 2))
    print(rotate_left(lista, -3))
    print(rotate_left(lista, 0))
    print('*=' * 30)

    print(split_even_odd(lista))
    print('*=' * 30)

    print(merge_sorted([3, 6, 9], [2, 4, 8]))
    print('*=' * 30)

    print(second_largest([4, 5, 22, 7, 10, 34, 4, 0, -1, 3]))
    print(second_largest([5, 5]))
    print('*=' * 30)


if __name__ == "__main__":
    main()





```

```python
def format_monomial(coef, exp):
    n = str(coef)
    if coef > 1:
        if exp == 0:
            n = f'+{coef}'
        elif exp == 1:
            n = f'+{coef}x'
        else:
            n = f'+{coef}x**{exp}'

    elif coef == 1:
        if exp == 0:
            n = '+1'
        elif exp == 1:
            n = f'+x'
        else:
            n = f'+x**{exp}'

    elif coef == 0:
        n = ''

    else:
        if exp == 0:
            n = f'{coef}'
        elif exp == 1:
            n = f'{coef}x'
        else:
            n = f'{coef}x**{exp}'

    return n
    
def format_polynomial(coefs):
    polinomio = ''

    for i in range(len(coefs) - 1, -1, -1):
        coef = coefs[i]                 
        exp = i    
        monomio = format_monomial(coef, exp)
        polinomio += monomio

    if polinomio and polinomio[0] == '+':
        polinomio = polinomio[1:]

    polinomio = polinomio.replace('+', ' + ').replace('-', ' - ')

    # Tras añadir los espacios, si el primer coeficiente es - algo al principio del polinomio aparecerá un espacio innecesario.
    # Entonces, si el primer caracter es un espacio en blanco, se suprime:
    if polinomio and polinomio[0] == ' ': 
        polinomio = polinomio[1:]

    return polinomio

def evaluate(coefs, x):
    if not coefs:
        return 0

    valor = 0
    for i, coef in enumerate(coefs):
        valor += coef * (x ** i)

    return valor
        
def derivate(coefs):
    if len(coefs) <= 1:
        return []
    derivada = []
    # coefs[1:] para ignorar la constante e 'i' empieza en 1
    for i, coef in enumerate(coefs[1:], 1):
        coef = coef * i
        derivada.append(coef)
    return derivada

def main():
    lista1 = [4.0, 1.0, 0.0, 2.0]
    lista2 = [-2.0, 5.5, 1.0, 0.0, -5.0]
    coefs = lista2
    exp = len(coefs) - 1
    x = 2
   
    for i in range(len(coefs) - 1, -1, -1):
        coef = coefs[i]
        n = format_monomial(coef, exp)
        print(n)
        exp -= 1
    
    print(format_polynomial(coefs))

    print(evaluate(coefs, x))

    print(derivate(coefs))

if __name__ == "__main__":
    main()












































```

```python
def imprimir_cadena(lista_cad, letra):
    for cadena in lista_cad:
        if cadena[0].lower() == letra.lower():
            print(cadena)
def contar_subcadena(lista_cad):
    subcadena = input("Introduce la subcadena que deseas buscar: ")
    contador = 0
    
    for cadena in lista_cad:
        if subcadena.lower() in cadena.lower():  
            contador += 1
            
    print(f"La subcadena '{subcadena}' aparece en {contador} cadena(s).")

def longitud_cadena(lista_cad):
    print('Cadena más larga:', max(lista_cad, key=len))
    print('Cadena más corta:', min(lista_cad, key=len))

def comparar_cadenas(lista_cad):
    encontrado = False
    for i in range(len(lista_cad)):
        for j in range(i + 1, len(lista_cad)):
            if lista_cad[i] is lista_cad[j]:
                print(f"Coincidencia de obejto: '{lista_cad[i]}'")
                encontrado = True
                
    if not encontrado:
        print("No hay ninguna cadena en la lista que comparta el mismo objeto en memoria.")

def mayor_que_10(lista_cad):
    encontrado = False
    for cadena in lista_cad:
        if len(cadena) > 10:
            print(f"'{cadena}' --> {len(cadena)} caracteres.")
            encontrado = True
            
    if not encontrado:
        print("Ninguna cadena tiene más de 10 caracteres.")

def menu(lista_cad):
    print("\n--- MENÚ DE OPCIONES ---")
    print("1. Imprima cada cadena que comience con una letra pedida al usuario.")
    print("2. Cuente cuántas cadenas contienen una subcadena pedida al usuario.")
    print("3. Encuentre la cadena más larga y la más corta de la lista.")
    print("4. Verificar si dos cadenas específicas de la lista son en realidad el mismo objeto.")
    print("5. Verificar si alguna cadena de la lista tiene una longitud mayor que 10.")
    print("6. Salir.")
    print("------------------------")
    
    print(lista_cad)
    opcion = int(input("\nSelecciona una opción (1 - 6): "))

    if opcion == 1:
        letra = input("Introduce letra: ")
        if len(letra) != 1:
            print('Solo se puede introducir una letra.')
        else:
            imprimir_cadena(lista_cad, letra)
    elif opcion == 2:
        contar_subcadena(lista_cad)
    elif opcion == 3:
        longitud_cadena(lista_cad)
    elif opcion == 4:
        comparar_cadenas(lista_cad)
    elif opcion == 5:
       mayor_que_10(lista_cad)
    elif opcion == 6:
        print("Saliendo...")
        return 6
    else:
        return False
        
def main():
    palabra = 'hola'
    lista_cad = [palabra, 'que tal', palabra]

    while True:
        cad = input("Escribe una frase (introduce 'Enter' para parar): ")
        if cad == "":
            break
        lista_cad.append(cad)

    if not lista_cad:
        print("No se ha introducido ninguna cadena.")
        return
    
    while True:
        resultado = menu(lista_cad)
        if resultado == 6:
            break
        elif resultado is False:
            print("El dato introducido es erróneo. Vuelve a intentarlo.")
      

if __name__ == "__main__":
    main()
```

```python
def is_anagram(word1, word2):
    # sorted(x): devuelve una nueva lista con sus elementos ordenados
    return sorted(word1.lower().replace(" ", "")) == sorted(word2.lower().replace(" ", ""))

def group_anagrams(words):
    lista_de_listas = []
    
    for word in words:
        encontrado = False
        for grupo in lista_de_listas:
            if is_anagram(word, grupo[0]):
                grupo.append(word)
                encontrado = True
                break
        
        if not encontrado:
            lista_de_listas.append([word])
            
    return lista_de_listas

def largest_group(groups):
    return [] if not groups else max(groups, key=len)

def find_anagrams(word, words):
    anagrams = []
    for w in words:
        if is_anagram(word, w) and w != word:
            anagrams.append(w)
    return anagrams

def main():
    words = ["roma", "amor", "pelo", "mora", "casa", "lope", "ramo"]

    print( 
        is_anagram("Roma", "amor"),"\n",
        is_anagram("casa", "caso"),"\n",
        is_anagram("dormitory", "dirty room"),"\n\n",
        
        group_anagrams(words),"\n","\n",
        
        largest_group(group_anagrams(words)),"\n\n",

        find_anagrams("amor", words),"\n",
        find_anagrams("amor", ["amor", "roma", "AMOR", "roma"]),"\n",
        find_anagrams("sol", words) )


if __name__ == "__main__":
    main()
```
